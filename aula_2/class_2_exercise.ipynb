# 📊 insight

```python
# Filtrando os dados de homens e mulheres 
homens = df[df['Sex'] == 0]  # Homens (Sex == 0)
mulheres = df[df['Sex'] == 1]  # Mulheres (Sex == 1)

# Pegando 400 amostras aleatórias de cada grupo
homens_400 = homens.sample(400, random_state=42)
mulheres_400 = mulheres.sample(400, random_state=42)

# Concatenando as amostras balanceadas
df_balanceado_400 = pd.concat([homens_400, mulheres_400])

# Calculando a taxa de sobrevivência para homens e mulheres no conjunto balanceado
taxa_sobrevivencia_400 = df_balanceado_400.groupby('Sex')['2urvived'].mean()

# Plotando o gráfico de barras de sexo
plt.bar([0, 1], taxa_sobrevivencia_400, color=['blue', 'pink'])
plt.xticks([0, 1], ['Homem', 'Mulher'])
plt.xlabel('Sexo')
plt.ylabel('Taxa de Sobrevivência')
plt.title('Taxa de Sobrevivência entre Homens e Mulheres (400 amostras balanceadas)')
plt.show()
```

```python
dados = df.copy()

dados['Age'] = pd.to_numeric(dados['Age'], errors='coerce')
dados['2urvived'] = pd.to_numeric(dados['2urvived'], errors='coerce')

dados = dados.dropna(subset=['Age', '2urvived'])

dados['Sexo'] = dados['Sex'].map({
    0: 'Homem',
    1: 'Mulher'
})

faixas = [0, 12, 18, 30, 45, 60, 100]
nomes_faixas = ['0-12', '13-18', '19-30', '31-45', '46-60', '61+']

dados['Faixa_Idade'] = pd.cut(dados['Age'],bins=faixas,labels=nomes_faixas,include_lowest=True)

resumo_idade = (
    dados.groupby(['Sexo', 'Faixa_Idade'], observed=True)
    .agg(
        total=('2urvived', 'count'),
        sobreviveram=('2urvived', 'sum'),
        taxa=('2urvived', 'mean'),
        idade_media=('Age', 'mean')
    )
    .reset_index()
)

resumo_idade['taxa_%'] = (resumo_idade['taxa'] * 100).round(1)
resumo_idade['idade_media'] = resumo_idade['idade_media'].round(1)

resumo_idade
```

### conclusão da Analise
Pela análise da taxa de sobrevivência por faixa etária, foi possível perceber que as mulheres tiveram maior chance de sobreviver do que os homens em praticamente todas as idades. Entre os homens, a maior sobrevivência ocorreu na faixa de 0 a 12 anos, com 42%, mostrando que crianças tiveram mais chances que homens jovens e adultos. Já entre as mulheres, as maiores taxas foram nas faixas de 31 a 45 anos, com 56,4%, e 13 a 18 anos, com 56,2%.

De forma geral, o gráfico mostra que a sobrevivência no Titanic não foi igual para todos: mulheres e crianças tiveram mais chances de sobreviver, enquanto os homens adultos apresentaram as menores taxas.
